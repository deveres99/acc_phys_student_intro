## Exercise 4 — Functions: converting electron-volts to joules

Physicists use two units of energy:
- **joule (J)**: the everyday unit. Lifting a small apple by 1 metre takes about 1 J.
- **electron-volt (eV)**: a *tiny* unit, perfect for single particles.

They're related by: $1\ \text{eV} = 1.602176634 \times 10^{-19}\ \text{J}$

So to go from eV to J you **multiply** by that number; to go from J to eV you **divide** by it.

### The Python you need:

A function can take inputs (called **parameters**) and `return` a result. It's good practice to add a **docstring** (a description in triple quotes) so others know what it does:
```python
def rectangle_area(width, height):
    """Return the area of a rectangle."""
    area = width * height
    return area

print(rectangle_area(3, 4))   # 12
```
Once written, a function can be used again and again. No copy-pasting formulas!
